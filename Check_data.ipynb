```python
import pandas as pd
```

```python
meta = pd.read_csv("metadata.csv")
print (len(meta))
```

```python
 print("Repeated IDs:", meta["ID"].duplicated().sum())
```

```python
meta.isna().sum()
```

```python
decades = []

for date in meta["Date"].dropna():
    year = int(str(date)[:4])
    decade = year // 10 * 10
    decades.append(decade)

print("Letters per decade:")
print(pd.Series(decades).value_counts().sort_index())

print("Number of letters per sender:")
print(meta["Sender"].value_counts())
```
